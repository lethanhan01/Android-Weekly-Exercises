```kotlin
fun main() {

    // ==========================================
    // 1. OPERATORS AND TYPES
    // ==========================================

    println("=== Operators and Types ===")

    println(1 + 1)
    println(53 - 3)
    println(50 / 10)
    println(1.0 / 2.0)
    println(2.0 * 3.5)

    println(1 / 2)
    println(1.0 / 2.0)

    println(6 * 50)
    println(6.0 * 50.0)
    println(6.0 * 50)

    println(2.times(3))
    println(3.5.plus(4))
    println(2.4.div(2))


    // ==========================================
    // 2. TYPE CONVERSION
    // ==========================================

    println("\n=== Type Conversion ===")

    val i: Int = 6
    val b1 = i.toByte()

    println(b1)

    val b2: Byte = 1

    val i4: Int = b2.toInt()
    val i5: String = b2.toString()
    val i6: Double = b2.toDouble()

    println(i4)
    println(i5)
    println(i6)


    // ==========================================
    // 3. NUMERIC LITERALS
    // ==========================================

    println("\n=== Numeric Literals ===")

    val oneMillion = 1_000_000
    val socialSecurityNumber = 999_99_9999L
    val hexBytes = 0xFF_EC_DE_5E
    val bytes = 0b11010010_01101001_10010100_10010010

    println(oneMillion)
    println(socialSecurityNumber)
    println(hexBytes)
    println(bytes)


    // ==========================================
    // 4. VARIABLES
    // ==========================================

    println("\n=== Variables ===")

    var fish = 1
    fish = 2

    println(fish)

    val aquarium = 1

    println(aquarium)

    var fishCount: Int = 12
    var lakes: Double = 2.5

    println(fishCount)
    println(lakes)


    // ==========================================
    // 5. STRINGS AND STRING TEMPLATES
    // ==========================================

    println("\n=== Strings ===")

    val numberOfFish = 5
    val numberOfPlants = 12

    println(
        "I have $numberOfFish fish" +
                " and $numberOfPlants plants"
    )

    println(
        "I have ${numberOfFish + numberOfPlants} fish and plants"
    )


    // ==========================================
    // 6. IF / ELSE
    // ==========================================

    println("\n=== Conditions ===")

    val fishInTank = 50
    val plantsInTank = 23

    if (fishInTank > plantsInTank) {
        println("Good ratio!")
    } else {
        println("Unhealthy ratio")
    }

    if (fishInTank in 1..100) {
        println(fishInTank)
    }

    if (fishInTank == 0) {
        println("Empty tank")
    } else if (fishInTank < 40) {
        println("Got fish!")
    } else {
        println("That's a lot of fish!")
    }


    // ==========================================
    // 7. WHEN
    // ==========================================

    println("\n=== When ===")

    when (fishInTank) {
        0 -> println("Empty tank")
        in 1..39 -> println("Got fish!")
        else -> println("That's a lot of fish!")
    }


    // ==========================================
    // 8. NULLABILITY
    // ==========================================

    println("\n=== Nullability ===")

    var marbles: Int? = null

    println(marbles)

    var fishFoodTreats: Int? = 6

    fishFoodTreats = fishFoodTreats?.dec()

    println(fishFoodTreats)

    fishFoodTreats = fishFoodTreats?.dec() ?: 0

    println(fishFoodTreats)


    // ==========================================
    // 9. LISTS
    // ==========================================

    println("\n=== Lists ===")

    val school = listOf(
        "mackerel",
        "trout",
        "halibut"
    )

    println(school)

    val myList = mutableListOf(
        "tuna",
        "salmon",
        "shark"
    )

    myList.remove("shark")

    println(myList)


    // ==========================================
    // 10. ARRAYS
    // ==========================================

    println("\n=== Arrays ===")

    val schoolArray = arrayOf(
        "shark",
        "salmon",
        "minnow"
    )

    println(
        java.util.Arrays.toString(schoolArray)
    )

    val mix = arrayOf("fish", 2)

    println(
        java.util.Arrays.toString(mix)
    )

    val numbers = intArrayOf(1, 2, 3)

    println(
        java.util.Arrays.toString(numbers)
    )

    val numbers3 = intArrayOf(4, 5, 6)

    val combined = numbers3 + numbers

    println(combined[5])


    // ==========================================
    // 11. NESTED LIST / ARRAY
    // ==========================================

    println("\n=== Nested Collections ===")

    val numbersNested = intArrayOf(1, 2, 3)

    val oceans = listOf(
        "Atlantic",
        "Pacific"
    )

    val oddList = listOf(
        numbersNested,
        oceans,
        "salmon"
    )

    println(oddList)


    // ==========================================
    // 12. ARRAY INITIALIZATION
    // ==========================================

    println("\n=== Array Initialization ===")

    val array = Array(5) {
        it * 2
    }

    println(
        java.util.Arrays.toString(array)
    )


    // ==========================================
    // 13. FOR LOOP
    // ==========================================

    println("\n=== For Loop ===")

    for (element in schoolArray) {
        print("$element ")
    }

    println()

    for ((index, element) in schoolArray.withIndex()) {
        println(
            "Item at $index is $element"
        )
    }


    // ==========================================
    // 14. RANGES
    // ==========================================

    println("\n=== Ranges ===")

    for (i in 1..5) {
        print(i)
    }

    println()

    for (i in 5 downTo 1) {
        print(i)
    }

    println()

    for (i in 3..6 step 2) {
        print(i)
    }

    println()

    for (i in 'd'..'g') {
        print(i)
    }

    println()


    // ==========================================
    // 15. WHILE / DO-WHILE / REPEAT
    // ==========================================

    println("\n=== Loops ===")

    var bubbles = 0

    while (bubbles < 50) {
        bubbles++
    }

    println(
        "$bubbles bubbles in the water"
    )

    do {
        bubbles--
    } while (bubbles > 50)

    println(
        "$bubbles bubbles in the water"
    )

    repeat(2) {
        println("A fish is swimming")
    }
}
```
